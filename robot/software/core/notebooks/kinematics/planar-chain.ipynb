# Chuỗi khớp phẳng — Ví dụ kỹ thuật

Ví dụ API Python tối giản, **không phải mô hình chân DUK**. Các góc là góc khớp tương đối (rad), chiều dài tính bằng cùng một đơn vị.


```python
from math import pi
from duk_core.modeling.planar import forward_chain
from duk_core.visualization.planar import draw_chain

lengths = [1.0, 0.5]
angles = [0.0, pi / 2]
points = forward_chain(lengths, angles)
points
```

```python
ax = draw_chain(points)
ax.set_title('Generic planar chain')
```
